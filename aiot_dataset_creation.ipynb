**Before inserting documents, your parsing code must:**

* Ensure loaded data are in coordination with "Sensor Selection Strategy" in the root `README.md`.
* Ensure the working sampling rate is **100 Hz**. Downsample any channel that is not natively at 100 Hz.

Each document in the MongoDB database should have the following schema (default configuration: hand/wrist IMU, accelerometer + gyroscope):

```json
{
  "_id": ObjectId("6984b3fa87abe7f4dff571aa"),
  "data": {
    "acc_x": ["array", "of", "values"],
    "acc_y": ["array", "of", "values"],
    "acc_z": ["array", "of", "values"],
    "gyr_x": ["array", "of", "values"],
    "gyr_y": ["array", "of", "values"],
    "gyr_z": ["array", "of", "values"]
  },
  "activity_id": 4,
  "activity_label": "walking",
  "subject": "101",
  "split": "Protocol",
  "imu_location": "hand",
  "sensor": "AccGyr",
  "sr": 100,
  "datetime": "MongoDB datetime object (it can be generated with the datetime.datetime.now() function)"
}
```

Field notes:

* `activity_id` is the integer label as it appears in the `.dat` file. `activity_label` is the human-readable name (see the activity table in the root `README.md`).
* `subject` is the file's subject identifier (e.g. `"101"` … `"109"`).
* `split` is `"Protocol"` or `"Optional"`, matching the source subdirectory.
* `imu_location` is `"hand"`, `"chest"`, or `"ankle"`. Use **one document per IMU location**; do not concatenate IMUs into a single document.
* `sensor` is `"Acc"`, `"Gyr"`, or `"AccGyr"`. If you later expand to the magnetometer, use `"AccGyrMag"` and add `mag_x`, `mag_y`, `mag_z` keys with the same axis convention.
* `sr` (sampling rate) **must be 100**. Downsample any channel that is not natively at 100 Hz before inserting.

If you are using only the gyroscope for an experiment, drop the `acc_*` keys and keep `gyr_x`, `gyr_y`, `gyr_z`. The same convention applies to the accelerometer-only case.

**Note: Be careful, the document is mandatory to have the aforementioned schema, in order to argue and proceed with the rest of the processes later on, in data engineering, plotting, etc.**

## Provide the code to parse the PAMAP2 `.dat` files and upload the documents to MongoDB

````python
# Parse the .dat files, handle NaN values, group contiguous activity blocks,
# remove activity 0, and upload full segments to MongoDB.

# Activity ID to label mapping based on PAMAP2 documentation
activity_map = {
    1: "lying", 2: "sitting", 3: "standing", 4: "walking",
    5: "running", 6: "cycling", 7: "Nordic walking",
    9: "watching TV", 10: "computer work", 11: "car driving",
    12: "ascending stairs", 13: "descending stairs",
    16: "vacuum cleaning", 17: "ironing", 18: "folding laundry",
    19: "house cleaning", 20: "playing soccer", 24: "rope jumping"
}

# PAMAP2 column positions for each IMU. We use the 16g accelerometer and gyroscope only.
imu_sensor_columns = {
    "hand": {
        "acc_x": 4, "acc_y": 5, "acc_z": 6,
        "gyr_x": 10, "gyr_y": 11, "gyr_z": 12,
    },
    "chest": {
        "acc_x": 21, "acc_y": 22, "acc_z": 23,
        "gyr_x": 27, "gyr_y": 28, "gyr_z": 29,
    },
    "ankle": {
        "acc_x": 38, "acc_y": 39, "acc_z": 40,
        "gyr_x": 44, "gyr_y": 45, "gyr_z": 46,
    },
}

delete_result = col.delete_many({}) # Clear the collection before inserting new data
print(f"Deleted existing documents: {delete_result.deleted_count}")

docs_to_insert = []

for split in splits:

    split_path = os.path.join(data_path, split)
    files_in_split = sorted([
        f for f in os.listdir(split_path)
        if f.endswith(".dat")
    ])

    for file in files_in_split:

        file_path = os.path.join(split_path, file)
        raw_df = pd.read_csv(file_path, sep=r"\s+", header=None)

        # Extract subject + split
        subject_id = os.path.splitext(file)[0]
        subject_id = "".join(ch for ch in subject_id if ch.isdigit())

        for imu_location, sensor_columns in imu_sensor_columns.items():
            # Keep the activity id and the selected IMU channels.
            df = pd.DataFrame({
                "activity_id": raw_df.iloc[:, 1].astype(int),
                **{name: raw_df.iloc[:, idx] for name, idx in sensor_columns.items()},
            })

            # Handle missing sensor values without using heart-rate/biometric columns.
            sensor_names = list(sensor_columns.keys())
            df[sensor_names] = df[sensor_names].replace([np.inf, -np.inf], np.nan)
            df[sensor_names] = df[sensor_names].interpolate(
                method="linear",
                limit_direction="both",
            )
            df = df.dropna(subset=sensor_names)

            # Segment before removing activity 0 so transition rows split activities.
            df["segment_id"] = df["activity_id"].ne(df["activity_id"].shift()).cumsum()

            nonzero_segment_index = 0
            for _, segment_df in df.groupby("segment_id", sort=False):
                activity_id = int(segment_df["activity_id"].iloc[0])

                if activity_id == 0:
                    continue

                doc = {
                    "data": {
                        "acc_x": segment_df["acc_x"].tolist(),
                        "acc_y": segment_df["acc_y"].tolist(),
                        "acc_z": segment_df["acc_z"].tolist(),
                        "gyr_x": segment_df["gyr_x"].tolist(),
                        "gyr_y": segment_df["gyr_y"].tolist(),
                        "gyr_z": segment_df["gyr_z"].tolist(),
                    },
                    "activity_id": activity_id,
                    "activity_label": activity_map.get(activity_id, "unknown"),
                    "subject": subject_id,
                    "split": split,
                    "imu_location": imu_location,
                    "sensor": "AccGyr",
                    "sr": 100, # sampling rate 100Hz
                    "datetime": datetime.now(),
                    "segment_length": int(len(segment_df)),
                    "segment_index": nonzero_segment_index,
                }

                docs_to_insert.append(doc)
                nonzero_segment_index += 1

if docs_to_insert:
    insert_result = col.insert_many(docs_to_insert)
    print(f"Inserted documents: {len(insert_result.inserted_ids)}")
else:
    print("No documents were inserted. Check the dataset path and activity labels.")

print("Documents by split and IMU location:")
for item in col.aggregate([
    {"$group": {"_id": {"split": "$split", "imu_location": "$imu_location"}, "count": {"$sum": 1}}},
    {"$sort": {"_id.split": 1, "_id.imu_location": 1}},
]):
    print(item)
print("Documents by activity:")
for item in col.aggregate([
    {"$group": {"_id": "$activity_label", "count": {"$sum": 1}}},
    {"$sort": {"_id": 1}},
]):
    print(item)
````
